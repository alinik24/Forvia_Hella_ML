```
import pandas as pd
```

```
final_df = pd.read_parquet(r"M:\Universität\Master\Semester2\RealWorld_ML_Problems\Data\data_hella_single_line\merged\final_merged_all.parquet")
```

```
# 1. Drop duplicate rows
final_df.drop_duplicates(inplace=True)
```

```
# 2. Drop duplicate columns
final_df = final_df.loc[:, ~final_df.T.duplicated()]
```

```
# 3. Convert all columns to float
for col in final_df.columns:
    if final_df[col].dtype == 'datetime64[ns]':
        # Convert datetime to UNIX timestamp (in float)
        final_df[col] = final_df[col].astype('int64') / 1e9
    elif final_df[col].dtype == 'object':
        # Ordinal encode text columns
        final_df[col] = final_df[col].astype('category').cat.codes.astype('float')
    else:
        # Try to cast directly to float
        try:
            final_df[col] = final_df[col].astype('float')
        except:
            # If conversion fails, mark it
            print(f"⚠️ Could not convert column: {col}")
```

```
# Final shape after processing
print("✅ Cleaned shape:", final_df.shape)
print("✅ Data types:", final_df.dtypes.value_counts())
```

```
# Save cleaned version
final_cleaned_path = "/path/to/output_dir/final_cleaned_float.parquet"
final_df.to_parquet(final_cleaned_path, index=False)
print("📦 Saved cleaned file to:", final_cleaned_path)
```
