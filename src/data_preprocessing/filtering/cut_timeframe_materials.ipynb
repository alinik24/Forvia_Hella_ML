```python
import os
import polars as pl
```

```python
base_path = r"M:\Universität\Master\Semester2\RealWorld_ML_Problems\Data\data_hella_single_line\filtered"
output_dir = r"M:\Universität\Master\Semester2\RealWorld_ML_Problems\Data\data_hella_single_line\filtered"

materials_file = os.path.join(base_path, "filtered_materials_encoded.parquet")
filtered_output_file = os.path.join(output_dir, "filtered_materials_encoded_time_cut.parquet")
```

```python
# Define target time window, this is the timewindow of the bookings and measurements merged dataset
start_time = "2025-03-01T01:21:20.773Z"
end_time = "2025-05-14T01:02:06.180Z"
```

```python
# Load Parquet with Polars
df = pl.read_parquet(materials_file)
```

```python
# Filter by 'created_at' time window
filtered = df.filter(
    (pl.col("created_at") >= pl.lit(start_time).str.strptime(pl.Datetime, strict=False)) &
    (pl.col("created_at") <= pl.lit(end_time).str.strptime(pl.Datetime, strict=False))
)
```

```python
# Filter by 'created_at' time window
filtered = df.filter(
    (pl.col("created_at") >= pl.lit(start_time).str.strptime(pl.Datetime, strict=False)) &
    (pl.col("created_at") <= pl.lit(end_time).str.strptime(pl.Datetime, strict=False))
)
```

```python
# Save the filtered dataset
filtered.write_parquet(filtered_output_file)
print(f"Filtered materials saved to: {filtered_output_file}")
```
